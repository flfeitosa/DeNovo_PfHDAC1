```python
import pandas as pd
```

```python
df= pd.read_csv('concat_datasets_highdiv.csv')
df
```

```python
# Sequential number per dataset (starting at 1), following the current row order
seq = df.groupby('Dataset').cumcount() + 1

# Build the id as "<Dataset>_<sequence>", e.g. CONCAT_1, ENAMINE_1
df['id'] = df['Dataset'].astype(str) + '_' + seq.astype(str)

# Move the id column to the front
df = df[['id'] + [c for c in df.columns if c != 'id']]
df
```

```python
df.to_csv('concat_datasets_highdiv_ids.csv', index=False)
```
